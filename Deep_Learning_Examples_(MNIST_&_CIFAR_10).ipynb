```
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import time

# --- Helper Function for Plotting History ---
def plot_history(history, model_name):
    """Plots training & validation accuracy and loss."""
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

    # Plot accuracy
    ax1.plot(history.history['accuracy'], label='Training Accuracy')
    ax1.plot(history.history['val_accuracy'], label='Validation Accuracy')
    ax1.set_title(f'{model_name} - Accuracy')
    ax1.set_xlabel('Epoch')
    ax1.set_ylabel('Accuracy')
    ax1.legend()
    ax1.grid(True)

    # Plot loss
    ax2.plot(history.history['loss'], label='Training Loss')
    ax2.plot(history.history['val_loss'], label='Validation Loss')
    ax2.set_title(f'{model_name} - Loss')
    ax2.set_xlabel('Epoch')
    ax2.set_ylabel('Loss')
    ax2.legend()
    ax2.grid(True)

    plt.tight_layout()
    plt.show()

# --- 1. MNIST Handwritten Digit Recognition (CNN) ---

print("\n--- Loading MNIST Dataset ---")
# Load the dataset directly from Keras
(x_train_mnist, y_train_mnist), (x_test_mnist, y_test_mnist) = keras.datasets.mnist.load_data()
print(f"MNIST Train shapes: X={x_train_mnist.shape}, y={y_train_mnist.shape}")
print(f"MNIST Test shapes: X={x_test_mnist.shape}, y={y_test_mnist.shape}")

# Preprocessing MNIST
# Scale images to the [0, 1] range
x_train_mnist = x_train_mnist.astype("float32") / 255.0
x_test_mnist = x_test_mnist.astype("float32") / 255.0

# Make sure images have shape (28, 28, 1) for CNN input
x_train_mnist = np.expand_dims(x_train_mnist, -1)
x_test_mnist = np.expand_dims(x_test_mnist, -1)
print(f"New MNIST Train shape: {x_train_mnist.shape}")
print(f"New MNIST Test shape: {x_test_mnist.shape}")

# Convert class vectors to binary class matrices (one-hot encoding)
num_classes_mnist = 10
y_train_mnist = keras.utils.to_categorical(y_train_mnist, num_classes_mnist)
y_test_mnist = keras.utils.to_categorical(y_test_mnist, num_classes_mnist)

# Build the CNN model for MNIST
mnist_model = keras.Sequential(
    [
        keras.Input(shape=(28, 28, 1)),
        # Convolutional Layer: Extracts features using filters
        # 32 filters, each 3x3 pixels. 'relu' activation introduces non-linearity.
        layers.Conv2D(32, kernel_size=(3, 3), activation="relu"),
        # MaxPooling Layer: Reduces spatial dimensions, makes features more robust
        layers.MaxPooling2D(pool_size=(2, 2)),
        # Another Conv layer
        layers.Conv2D(64, kernel_size=(3, 3), activation="relu"),
        # Another MaxPooling layer
        layers.MaxPooling2D(pool_size=(2, 2)),
        # Flatten Layer: Converts 2D feature maps to a 1D vector for Dense layers
        layers.Flatten(),
        # Dropout Layer: Regularization technique to prevent overfitting
        layers.Dropout(0.5),
        # Dense Layer (Output): Fully connected layer with units = number of classes
        # 'softmax' activation outputs probabilities for each class
        layers.Dense(num_classes_mnist, activation="softmax"),
    ]
)

mnist_model.summary()

# Compile the MNIST model
# Adam optimizer is a common choice.
# 'categorical_crossentropy' is standard loss for multi-class classification with one-hot labels.
mnist_model.compile(loss="categorical_crossentropy", optimizer="adam", metrics=["accuracy"])

# Train the MNIST model
batch_size_mnist = 128
epochs_mnist = 15 # Increase epochs for potentially better results
print("\n--- Training MNIST Model ---")
start_time = time.time()
history_mnist = mnist_model.fit(
    x_train_mnist,
    y_train_mnist,
    batch_size=batch_size_mnist,
    epochs=epochs_mnist,
    validation_split=0.1 # Use 10% of training data for validation during training
)
end_time = time.time()
print(f"MNIST Training Time: {end_time - start_time:.2f} seconds")

# Evaluate the MNIST model
print("\n--- Evaluating MNIST Model ---")
score_mnist = mnist_model.evaluate(x_test_mnist, y_test_mnist, verbose=0)
print(f"Test loss: {score_mnist[0]:.4f}")
print(f"Test accuracy: {score_mnist[1]:.4f}")

# Plot training history
plot_history(history_mnist, "MNIST CNN")


# --- 2. CIFAR-10 Image Classification (CNN) ---

print("\n\n--- Loading CIFAR-10 Dataset ---")
# Load the dataset directly from Keras
(x_train_cifar, y_train_cifar), (x_test_cifar, y_test_cifar) = keras.datasets.cifar10.load_data()
print(f"CIFAR-10 Train shapes: X={x_train_cifar.shape}, y={y_train_cifar.shape}")
print(f"CIFAR-10 Test shapes: X={x_test_cifar.shape}, y={y_test_cifar.shape}")
cifar_class_names = ['airplane', 'automobile', 'bird', 'cat', 'deer',
                   'dog', 'frog', 'horse', 'ship', 'truck']

# Preprocessing CIFAR-10
# Scale images to the [0, 1] range
x_train_cifar = x_train_cifar.astype("float32") / 255.0
x_test_cifar = x_test_cifar.astype("float32") / 255.0

# Convert class vectors to binary class matrices (one-hot encoding)
num_classes_cifar = 10
y_train_cifar = keras.utils.to_categorical(y_train_cifar, num_classes_cifar)
y_test_cifar = keras.utils.to_categorical(y_test_cifar, num_classes_cifar)

# Build a slightly deeper CNN model for CIFAR-10
cifar_model = keras.Sequential(
    [
        keras.Input(shape=(32, 32, 3)), # Input shape for CIFAR-10 (3 color channels)
        # Block 1
        layers.Conv2D(32, kernel_size=(3, 3), padding="same", activation="relu"),
        layers.BatchNormalization(), # Helps stabilize and speed up training
        layers.Conv2D(32, kernel_size=(3, 3), padding="same", activation="relu"),
        layers.BatchNormalization(),
        layers.MaxPooling2D(pool_size=(2, 2)),
        layers.Dropout(0.25), # Dropout after pooling

        # Block 2
        layers.Conv2D(64, kernel_size=(3, 3), padding="same", activation="relu"),
        layers.BatchNormalization(),
        layers.Conv2D(64, kernel_size=(3, 3), padding="same", activation="relu"),
        layers.BatchNormalization(),
        layers.MaxPooling2D(pool_size=(2, 2)),
        layers.Dropout(0.25),

        # Block 3
        layers.Conv2D(128, kernel_size=(3, 3), padding="same", activation="relu"),
        layers.BatchNormalization(),
        layers.Conv2D(128, kernel_size=(3, 3), padding="same", activation="relu"),
        layers.BatchNormalization(),
        layers.MaxPooling2D(pool_size=(2, 2)),
        layers.Dropout(0.25),

        # Flatten and Dense layers
        layers.Flatten(),
        layers.Dropout(0.4), # Increased dropout before final dense layer
        layers.Dense(512, activation='relu'), # Add a larger dense layer
        layers.BatchNormalization(),
        layers.Dropout(0.4),
        layers.Dense(num_classes_cifar, activation="softmax"), # Output layer
    ]
)

cifar_model.summary()

# Compile the CIFAR-10 model
# Using Adam optimizer with a potentially smaller learning rate
optimizer_cifar = tf.keras.optimizers.Adam(learning_rate=0.001)
cifar_model.compile(loss="categorical_crossentropy", optimizer=optimizer_cifar, metrics=["accuracy"])

# Train the CIFAR-10 model
batch_size_cifar = 64 # Smaller batch size might be better for CIFAR-10
epochs_cifar = 30 # CIFAR-10 often requires more epochs
print("\n--- Training CIFAR-10 Model ---")
start_time = time.time()
# Consider adding data augmentation here for better results in a real scenario
history_cifar = cifar_model.fit(
    x_train_cifar,
    y_train_cifar,
    batch_size=batch_size_cifar,
    epochs=epochs_cifar,
    validation_split=0.1
)
end_time = time.time()
print(f"CIFAR-10 Training Time: {end_time - start_time:.2f} seconds")

# Evaluate the CIFAR-10 model
print("\n--- Evaluating CIFAR-10 Model ---")
score_cifar = cifar_model.evaluate(x_test_cifar, y_test_cifar, verbose=0)
print(f"Test loss: {score_cifar[0]:.4f}")
print(f"Test accuracy: {score_cifar[1]:.4f}")

# Plot training history
plot_history(history_cifar, "CIFAR-10 CNN")

print("\n--- Deep Learning Examples Complete ---")
```
